```python
#empty set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_set1={1,2,3,4,56,5,4,5,6,56}
print(my_set1)
```

```python
#basic set operation
#adding and removing element in set

my_set1.add(99)
print(my_set1)
```

```python
my_set1.add(69)
print(my_set1)
```

```python
my_set1.remove(56)
print(my_set1)
```

```python
my_set1.discard(11)
print(my_set1)
# here the discard do when the remove element is present or not if it doesn't have the element it won't throw error
```

```python
# pop method
removed_element=my_set1.pop()
print(removed_element)
print(my_set1)
```

```python
#clear the whole set
my_set1.clear()
print(my_set1)
```

```python
#mathematical operation sets

set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

#union 
union_set= set1.union(set2)
print(union_set)
```

```python
difference_set= set1.difference(set2)
print(difference_set)
```

```python
#symetric differene - unique elements are present and common elements are removed
set1.symmetric_difference(set2)
```

```python
set1={1,2,3,4}
set2={3,4,5,6}
print(set1.issubset(set2))
print(set1.issuperset(set2))
```
