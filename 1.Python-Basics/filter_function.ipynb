```python
#filter with dictionary
def get_age(person):
    return person['age']>=25
people=[
    {"name":"Arun","age":25},
    {"name":"karthik","age":20},
    {"name":"kumar","age":27}
]


greater_age=list(filter(get_age,people))
print(greater_age)
```
