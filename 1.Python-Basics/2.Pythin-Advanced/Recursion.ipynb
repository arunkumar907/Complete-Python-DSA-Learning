```python
#Check if array is sorted

def checkSorted(l1):
    if(len(l1)==0 or len(l1)==1):
        return True

    ans =checkSorted(l1[l1:])

    if(l1[0] < l1[1]):
        return ans
    else:
        return False

def checkedSorted2(l1):
    if (len(l1) == 0 or len(l1)==1):
        return True

    if (l1[0] >= l1[1]):
        return False

    return  checkedSorted2(l1[l1:])

l3=[i for i in range (990,950,-1)]

print(l3)
print(checkedSorted2(l3))
```

```python
# Sum of Array 

def sumArray(l1):

    if (len(l1) ==0):
        return 0

    sumOfLeftOverArray = sumArray(l1[1:])

    ans = sumOfLeftOverArray+l1[0]
    return ans

def sumArray2(l1):
    if(len(l1)==0):
        return 0

    ans = l1[0] + sumArray(l1[1:])
    return ans

def sumArray_Tail(l1,accumulator=0):
    if(len(l1)==0):
        return accumulator
    accumulator += l1[0]

    return sumArray_Tail(l1[1:],accumulator)


print(sumArray([1,2,3,4,5,6]))
print(sumArray([]))
print(sumArray2([10,20,30]))
print(sumArray2([]))
print(sumArray_Tail([100,200]))
print(sumArray_Tail([]))
```

```python
# First index of an element in list 

def firstIndexOfanElement(l1,x):
    #Base case
    if (len(l1)==0):
        return -1 
    
    if(l1[0] == x):
        return 0

    ansFromRecursion = firstIndexOfanElement(l1[1:],x)

    if(ansFromRecursion == -1):
        return ansFromRecursion
    else:
        return ansFromRecursion +1

def lastIndexOfanElement(l1,x):
    if (len(l1)==0):
        
        return -1 
        
    if(l1[-1] == x):
        return len(l1) - 1

    
    return lastIndexOfanElement(l1[:-1],x)
    
    

print(firstIndexOfanElement([3,5,6,7,2,1,3],3))
print(firstIndexOfanElement([3,5,6,7,2,1,3],10))
print(firstIndexOfanElement([3,5,6,7,2,1,3],2))

print(lastIndexOfanElement([3,5,6,7,2,1,3],3))
print(lastIndexOfanElement([3,5,6,7,2,1,10,5],5))

```

```python
#factorial number

def factorial(n):

    if n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)

factorial(4)    
```
