```python
# 1.Right angled triangle 

n=4

for i in range (1, n+1):
    for j in range (i):
        print("*",end="")
    print()
```

```python
# Inverted angle

n=5

for i in range (1,n+1):
    for j in range(n-i+1):
        print("*", end =" ")
    print()    
```

```python
#square of N Side

n=4

for i in range(n):
    for j in range(n):
        print("*", end=" ")
    print()    
```

```python
# Row and column - Rectange Pattern

row=int(input())
column=int(input())

for i in range(row):
    for j in range(column):
        print("*", end=" " )
    print()        
```

```python
# Pyramid pattern

n=5

for i in range(1, n+1):
    for j in range(n-i):
        print(" ", end="")
    for j in range (2* i-1):
        print("*", end="")    
    print()    
```

```python
#inverted
n=5

for i in range(1, n+1):
    for j in range(i-1):
        print(" ", end="")
    for j in range (2*(n-i)+1):
        print("*", end="")    
    print()   
```
