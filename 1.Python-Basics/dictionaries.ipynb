```python
student={"Name":"Arun","Age":27,"Grade":"A"}
print(student)
print(type(student))
```

```python
# acessing dictionary element
print(student['Grade'])
print(student['Age'])
# print(student['Last_name']) -- If we use this it'll throw an error rather we can use get function and it'll show None value

#acessing with get method
print(student.get('Grade'))
print(student.get('Last_name'))
```

```python
# modifying dictionary elements

student["Age"]=25
print(student)
student["Address"]='217,chennai'
print(student)
```

```python
del student['Grade']
```

```python
print(student)
```

```python
# Dictionary methods

keys=student.keys()
print(keys)
values=student.values()
print(values)

```

```python
items=student.items()
print(items)
```

```python
# normal  copy
 ## Here the changes will affect in both dictionry
student_copy=student
print(student)
print(student_copy)
student["Name"] = "Arunk"
print(student_copy)
print(student)
```

```python
# Shallow copy

student_copy1=student.copy()
print(student)
student["Name"]="Arunkumar"
print(student_copy1)
print(student)
```

```python
#iterating over dictionaries -- loops

for keys in student.keys():
    print(keys)
```

```python
for values in student.values():
    print(values)
```

```python
# iterate over key value pairs

for keys,values in student.items():
    print(f"{keys}:{values}")
```

```python
#Nested Dictionry
students={
    "student1":{"Name":"Arun","Age":27},
    "student2":{"Name":"IronMan","Age":35}
}
print(students)
```

```python
#Access Nested Dictionaries

print(students)
print(students["student2"]["Name"])
print(students["student2"]["Age"])
```

```python
# dictionary comprehension
squares= {x:x**2 for x in range(5)}
print(squares)
```

```python
#even only
even={x:x**2 for x in range(10) if x%2==0}
print(even)
```

```python
#usecases

numbers=[1,2,3,5,2,2,3,3,4,3]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)        
```

```python
#merge two dictionary into one

dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```
