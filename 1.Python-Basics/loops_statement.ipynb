```python
#while loop - while loop continues to execute as long as the condition is true
count=0
while count<=5:
    print(count)
    count=count+1
```

```python
#loop control statement


#break - statement  exit the loop permaturely

for i in range(10):
    if i==8:
        break
    print(i)

```

```python
#continue - statement skips the current iteration and continue with the next

for i in range(1,10):
    if i%2==0:
        continue
    print(i)
```

```python
#pass - is null statement it does nothing

for i in range(5):
    if i ==3:
        pass
    print(i)
```

```python
# nested loop statement - loop inside the loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
# sum of n natural numbers

n=10
sum=0
count=0

while count<=n:
    sum=sum+count
    count=count+1
print("Sum of 10 natural number:",sum)    
```

```python
# example - prime number between 1 and 100

for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)

```
