```python
#Accessing elements in list

fruits=["Apple","Banana","Cheery","Pineapple","Mango"]
```

```python
print(fruits[1])
print(fruits[2])
print(fruits[0])
print(fruits[-2])
```

```python
print(fruits[1:])
print(fruits[-1:-3])
```

```python
#modifying list elements
print(fruits[0:])
```

```python
print(fruits[2])
fruits[2]="Straberry"
print(fruits[2])
print(fruits[0:])
```

```python
#list methods

fruits.append("Orange")
print(fruits)
```

```python
fruits.insert(1,"Bananna")
print(fruits)
```

```python
#remove and return the last element

#pop - remove the last element
print(fruits)
poped_fruits=fruits.pop()
print(poped_fruits)
print(fruits)
```

```python
index=fruits.index('Orange')
print(index)
```

```python
#slicing
number=[1,2,3,4,5,6,7,8,9]
print(number[1:3])
print(number[1:])
print(number[:5])
print(number[:0])
print(number[:-1])
print(number[::2])
print(number[::-1])
```

```python
#iterrating with list

for index,number in enumerate(number):
    print(index,number)
```

```python
#list comprehension

list=[]

for x in range(10):
    list.append(x**2)
print(list)

[x**2 for x in range(10)]
```

```python
#Basic list comprehension

square= [num**2 for num in range(10)]
print(square)
```

```python
#list comprehension with conditional 

list=[]

for i in range(10):
    if i%2==0:
        list.append(i)
print(list)
```

```python
even_number=[num for num in range(10) if num%2==0]
print(even_number)
```

```python
# nested list comprehension

list1=[1,2,3,4,5]
list2=['a','b','c','d','e']
pair=[[i,j]for i in list1 for j in list2]
print(pair)
```

```python
# list comprehension with function call

words=["Arun","Kumar","BCA","Bangalore","India","USA"]
length=[len(word) for word in words]
print(length)
```

```python
wordss=["kk"]
lenn=[len(word) for word in wordss]
print(lenn)
```
