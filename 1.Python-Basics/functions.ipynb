```python
def even_or_odd(num):
    if num%2==0:
        print("The Given number",num,"is Even")
    else:
        print("The Given number",num,"is odd")
           
```

```python
#variable length Arguments

#positional and key word arguments

def print_numbers(*args):
    for number in args:
        print(number)

        
```

```python
print_numbers(1,2,3,4,5,6,7,"Arun")
```
