```python
# File Handling 

# read a whole file

with open('sample.txt',mode='r')as file:
    text=file.read()
    print(text)
```

```python
# read a file in Line by line

with open('sample.txt',mode='r') as file:
    for line in file:
        print(line.strip())
    

```

```python
#writing a file inside a file(Overwriting)


with open ('sample.txt', mode='w')as file:
    file.write('hello worlds! \n ')
    file.write("here is the new line \n")
```

```python
#writing a file inside a file(with out Overwriting)

with open('sample.txt', mode='a') as file:
    file.write("Append  the new line without over\n")
    file.write("ANd the next line")

```

```python
# Writing a list of lines to a file

lines=['First Line \n','Second Line \n','Third Line \n']
with open('sample.txt','a')as file:
    file.writelines(lines)

```

```python
#Binary files

data = b'\x50\x79\x74\x68\x6f\x6e\x20\x43\x6f\x75\x72\x73\x65\x20'
data1 = b'\x20\x41\x72\x75\x6e\x20\x4b\x75\x6d\x61\x72\x20'

with open ('sample.txt','wb') as file:
    file.write(data)
    file.write(data1)

```

```python
# read a binary file

with open('sample.txt','rb') as file:
    content=file.read()
    print(content)
```

```python
# read the content from a source file and write to a destination file
with open ('sample.txt','r') as source_file:
    content = source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)    
 
```

```python
# raed the text file and count the number of lines, words and charater

def count_text_file(file_path):
    with open (file_path,'r') as file:
        lines=file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(lines) for line in lines)
    return line_count,word_count,char_count

file_path="sample.txt"
lines,words,characters= count_text_file(file_path)
print(f"'Lines: {lines}, Words: {words}, Characters: {characters}")    


```

```python
# Writing and reading a file at same time

with open ('sample.txt','w+') as file:
    file.write("hello World \n")
    file.write("This is a new line \n")
    file.write("This is a writing and reading a new line file\n")

    #Move the file cursor to the begining

    file.seek(0)

    #read the content of the file
    content1=file.read()
    print(content1)
```
