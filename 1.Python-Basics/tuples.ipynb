```python
#packing and unpacking

packed_tuples=1,"Hello",3.16
print(packed_tuples)
```

```python
#unpacking tuple 
a,b,c=packed_tuples
print(a)
print(b)
print(c)
```

```python
#unpacking with star *

numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
# nested tuples
lst=[[1,2,3,4],[5,6,7,8],[1,"hello",3.14,c]]
print(lst[0])
print(lst[1][0:3])
```

```python
nested_tuple = ((1,2,3),(4,5,6),(1,"Hello",3.16,c),("a","b","c"))
print(nested_tuple[1])
print(nested_tuple[3][2])
```

```python
#iterating over nested tuples

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end='')
    print()        
```
