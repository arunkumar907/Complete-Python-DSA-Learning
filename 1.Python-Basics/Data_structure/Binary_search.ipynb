```python
# Binary Search - It should be sorted form and Time complexity is [O(log(n))]

arr=[10,20,25,30,25,50,55,70]
target = 70

start=0
end=len(arr)-1

while(start<=end):
    mid=((start+end))//2
    
    if (arr[mid] == target):
        print("Target found:",mid)
        break
    elif (arr[mid] < target):
        start=mid+1
    elif (arr[mid] > target):
        end=mid-1
else:
        print("No data found")        

```
