```python
class Stack:
    def __init__(self):
        self.items=[]

    def push(self,value):
        self.items.append(value)
        print(f"Pushed Value: {value} --> Stack: {self.items}")

    def pop(self):
        value=self.items.pop()
        print(f"Removed value: {value} --> Stack: {self.items}")        

    def peek(self):
        return self.items[-1]

    def is_empty(self):
        return len(self.items) ==0

    def size(self):
        return len(self.items)
    
    

array=[]    
s=Stack()
s.push(10)
s.push(20)
s.push(30)
s.pop()

print("\nlast Element is ", s.peek() )
print("\n isempty? ",s.is_empty())
print("\n Size:", s.size())
```

```python
# Reverese a string in stack

original="Hello"
def reverse (arr):
    stack=[]

    for item in arr:
        stack.append(item)

    reversed_str=""
    while stack:
        reversed_str += stack.pop()
    return reversed_str

print("Original:", original)
print("Reversed string :", reverse(original))
```
