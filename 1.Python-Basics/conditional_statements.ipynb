```python
#Nested condition stateent
#tell me given number is even or odd or negative number

num = float(input("Enter the number:"))

if num<0:
    print("Given Number",num,"is Negative")
elif num%2==0:
    print("Given number",num,"is even")  
else:
    print("Given number",num,"is odd" )
```

```python
#usecases

#deterine the given year is leap yaer or not

year=int(input("Enter the year:"))
if year%4==0:
    if year%100==0:
        if year%400==0:
            print("The Given",year,"is a leap year")
        else:
            print("The Given",year,"is not a leap year") 
    else:
        print("The Given",year,"is  a leap year") 
else:
    print("The Given",year,"is not a leap year")         
```
