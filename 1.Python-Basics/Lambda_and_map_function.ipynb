```python
#map()

numbers=[1,2,3,4,5]

def sqaure(number):
    return number**2

sqaure(2)
```

```python
list(map(lambda x:x**2,numbers))
```

```python
numbers=[1,2,3,4]
def sqaure(x):
    return x**2

list(map(sqaure,numbers))
```

```python
#map multiple iterations

numbers1=[1,2,3,4]
numbers2=[5,6,7,8]

added_number=list(map(lambda x,y:x+y,numbers1,numbers2))
print(added_number)
```

```python
# convert string into numbers           
str_numbers=['1','2','3']
int_numbers=list(map( int,str_numbers))

print(int_numbers)
```

```python
#convert all lower case into upper case
words=["Arun","kumar","mca","chennai"]
upper_words=list(map(str.upper,words))
print(upper_words)
```
