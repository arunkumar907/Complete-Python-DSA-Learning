```python
#1.To -do list

to_do_list=["Buy Groceries","Pay Bills","Schedule for meeting","clean the house"]

##Adding the lists

to_do_list.append("Go for a Run")
to_do_list.append("Car wash")
to_do_list.append("Service Reminder")

#removing a completed task

to_do_list.remove("clean the house")
to_do_list.remove("Car wash")

#check if the task is done

if "Pay Bills" in to_do_list:
    print("Don't forget to do the task")

print("Remaining todo List are:")
for task in to_do_list:
    print(f"-{task}")
```

```python
#Example 2: Organizing student Grades

#organising the grades
grades=[85,90,95,75,80,90]

#Adding a new grade
grades.append(92)
grades.append(89)

#calculating the average

average_grade=sum(grades)/len(grades)
print(f"Average Grade:{average_grade}")

#calculating the lowest and highest grade

highest_grade =max(grades)
lowest_grade = min(grades)
print(f"Highest Grade:{highest_grade}")
print(f"Lowest Grade:{lowest_grade}")
```

```python
# Example 3: Manage an inventory

inventory=["Apple","Banana","Cheery","Orange","Strawberry"]

#adding a item in inventory

inventory.append("Pineapple")
inventory.append("Watermelon")
inventory.append("Gauva")

#removing an item which are out of stock
inventory.remove("Cheery")

#checking if the item in stock or not

item ="Cheery"
if item in inventory:
    print(f"{item} is available in Stock")
else:
    print(f"{item} is not available. Need to be Restocked")
    
print("Remaining Stocks are:")
for balance in inventory:
    print(f"-{balance}")        
```

```python
# Example:4 -- Collecting a feedback 

feedback=["Excellent service","great work","Very satisfied","Amazing experience","could be better"]

#adding a feedback
feedback.append("Not happy with the service, Need to be improve in support")

#counting some feedback

positive_feedback_count= sum(1 for comment in feedback if "great" in comment.lower() or "excellent" in comment.lower())
print(f"Positive Feedback count is {positive_feedback_count}")

#Printing all feedback
print("All user feedbacks are:")
for comment in feedback:
    print(f"-{comment}")
```
